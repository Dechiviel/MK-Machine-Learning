# Find-S Algorithm for Enjoy Sports dataset

```python
import pandas as pd

def find_s(instances):
    features, labels = instances
    if len(features) == 0:
        return []
    hypothesis = [None] * len(features[0])
    for i in range(0, len(features)):
        if labels[i]==1:
            attributes = features[i]
            for j in range(0, len(attributes)):
                if hypothesis[j] == None:
                    hypothesis[j] = attributes[j]
                elif hypothesis[j] != attributes[j]:
                    hypothesis[j] = "?"
            print(hypothesis)
    return hypothesis

if __name__ == '__main__':
    data = pd.read_csv('enjoy_sport.csv', sep='\t')
    hypothesis = find_s([data[['Sky', 'Temp', 'Humid', 'Wind', 'Water', 'Forecst']].values, data['EnjoySpt'].apply(lambda x: 1 if x == "Yes" else 0).values])
    print(hypothesis)
```
